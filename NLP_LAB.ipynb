```python
### exp1
import re

text = "My phone number is 9876543210"

if re.search(r"\d{10}", text):
    print("Number found")

match = re.findall(r"\d+", text)
print(match)
```

```python
### exp 2
def fsa(s):
    state = 0
    for ch in s:
        if state == 0:
            if ch == 'a':
                state = 1
        elif state == 1:
            if ch == 'b':
                state = 2
            elif ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0
    if state == 2:
        print("Accepted")
    else:
        print("Rejected")

fsa("aaab")
```

```python
### exp 3
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

words = ["running", "better", "cars"]

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

```python
### exp 4
def plural(word):
    if word.endswith("y"):
        return word[:-1] + "ies"
    elif word.endswith(("s", "x", "z", "ch", "sh")):
        return word + "es"
    else:
        return word + "s"

words = ["cat", "bus", "baby", "box"]

for w in words:
    print(plural(w))
```

```python
## exp 5
import nltk
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = ["playing", "played", "plays", "running"]

for word in words:
    print(word, "->", ps.stem(word))
```

```python
### exp 6
import random

text = "I like python I like coding python is easy"

words = text.split()

bigrams = {}

for i in range(len(words)-1):
    if words[i] not in bigrams:
        bigrams[words[i]] = []
    bigrams[words[i]].append(words[i+1])

word = "I"

result = [word]

for i in range(10):
    if word in bigrams:
        word = random.choice(bigrams[word])
        result.append(word)
    else:
        break

print(" ".join(result))
```

```python
### exp 7
import nltk

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng') # Corrected typo from ltk.download to nltk.download
nltk.download('punkt_tab') # Added to resolve the LookupError

text = "Python is easy to learn"

words = nltk.word_tokenize(text)

print(nltk.pos_tag(words))
```

```python
### edxp 8
import random

words = ["I", "play", "cricket"]

tags = {
    "I": ["PRP"],
    "play": ["VB", "NN"],
    "cricket": ["NN"]
}

for word in words:
    print(word, random.choice(tags[word]))
```

```python
## exp9
import nltk

patterns = [
    (r'.*ing$', 'VBG'),
    (r'.*ed$', 'VBD'),
    (r'.*ly$', 'RB'),
    (r'.*s$', 'NNS'),
    (r'.*', 'NN')
]

tagger = nltk.RegexpTagger(patterns)

words = ["playing", "worked", "slowly", "cats", "apple"]

print(tagger.tag(words))
```

```python
### epx 10

words = ["I", "can", "fish"]

tags = ["NN", "NN", "NN"]

for i in range(len(words)):
    if words[i] == "I":
        tags[i] = "PRP"
    elif words[i] == "can":
        tags[i] = "MD"

for w, t in zip(words, tags):
    print(w, t)
```
