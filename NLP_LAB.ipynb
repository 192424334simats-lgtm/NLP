```python
### exp 6
import random

text = "I like python I like coding python is easy"

words = text.split()

bigrams = {}

for i in range(len(words)-1):
    if words[i] not in bigrams:
        bigrams[words[i]] = []
    bigrams[words[i]].append(words[i+1])

word = "I"

result = [word]

for i in range(10):
    if word in bigrams:
        word = random.choice(bigrams[word])
        result.append(word)
    else:
        break

print(" ".join(result))
```

```python
## exp9
import nltk

patterns = [
    (r'.*ing$', 'VBG'),
    (r'.*ed$', 'VBD'),
    (r'.*ly$', 'RB'),
    (r'.*s$', 'NNS'),
    (r'.*', 'NN')
]

tagger = nltk.RegexpTagger(patterns)

words = ["playing", "worked", "slowly", "cats", "apple"]

print(tagger.tag(words))
```

```python
### epx 10

words = ["I", "can", "fish"]

tags = ["NN", "NN", "NN"]

for i in range(len(words)):
    if words[i] == "I":
        tags[i] = "PRP"
    elif words[i] == "can":
        tags[i] = "MD"

for w, t in zip(words, tags):
    print(w, t)
```

### 12. Implement an Earley parser for context-free grammars using a simple python program.

```python
sentence = ["the", "cat"]

grammar = {
    "S": ["the cat"]
}

text = " ".join(sentence)

if text in grammar["S"]:
    print("Accepted")
else:
    print("Rejected")
```

### 19. Create a program for word sense disambiguation using the Lesk algorithm using python.

```python
### exp 23
text = "Ravi likes football. Football is his favorite game. He plays football daily."

sentences = text.split(".")
score = 0

for i in range(1, len(sentences)):
    a = set(sentences[i-1].lower().split())
    b = set(sentences[i].lower().split())

    if a & b:
        score += 1

print("Coherence Score:", score)
```

```python
### exp 24

dialog = input("Enter a sentence: ").lower()

if "?" in dialog:
    act = "Question"
elif any(word in dialog for word in ["hello", "hi", "hey"]):
    act = "Greeting"
elif any(word in dialog for word in ["thanks", "thank you"]):
    act = "Thanking"
elif any(word in dialog for word in ["bye", "goodbye"]):
    act = "Goodbye"
else:
    act = "Statement"

print("Dialog Act:", act)
```

```python
### exp26

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load tokenizer and model directly
tokenizer = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-en-fr")
model = AutoModelForSeq2SeqLM.from_pretrained("Helsinki-NLP/opus-mt-en-fr")

text = input("Enter English text: ")

# Encode the input text
input_ids = tokenizer(text, return_tensors="pt").input_ids

# Generate translation
outputs = model.generate(input_ids)

# Decode the output
translated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("French:", translated_text)
```
