```python
### exp 2
def fsa(s):
    state = 0
    for ch in s:
        if state == 0:
            if ch == 'a':
                state = 1
        elif state == 1:
            if ch == 'b':
                state = 2
            elif ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0
    if state == 2:
        print("Accepted")
    else:
        print("Rejected")

fsa("aaab")
```

```python
### exp 3
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

words = ["running", "better", "cars"]

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

```python
### exp 4
def plural(word):
    if word.endswith("y"):
        return word[:-1] + "ies"
    elif word.endswith(("s", "x", "z", "ch", "sh")):
        return word + "es"
    else:
        return word + "s"

words = ["cat", "bus", "baby", "box"]

for w in words:
    print(plural(w))
```

```python
### exp 6
import random

text = "I like python I like coding python is easy"

words = text.split()

bigrams = {}

for i in range(len(words)-1):
    if words[i] not in bigrams:
        bigrams[words[i]] = []
    bigrams[words[i]].append(words[i+1])

word = "I"

result = [word]

for i in range(10):
    if word in bigrams:
        word = random.choice(bigrams[word])
        result.append(word)
    else:
        break

print(" ".join(result))
```

```python
## exp9
import nltk

patterns = [
    (r'.*ing$', 'VBG'),
    (r'.*ed$', 'VBD'),
    (r'.*ly$', 'RB'),
    (r'.*s$', 'NNS'),
    (r'.*', 'NN')
]

tagger = nltk.RegexpTagger(patterns)

words = ["playing", "worked", "slowly", "cats", "apple"]

print(tagger.tag(words))
```

```python
### epx 10

words = ["I", "can", "fish"]

tags = ["NN", "NN", "NN"]

for i in range(len(words)):
    if words[i] == "I":
        tags[i] = "PRP"
    elif words[i] == "can":
        tags[i] = "MD"

for w, t in zip(words, tags):
    print(w, t)
```

### 12. Implement an Earley parser for context-free grammars using a simple python program.

```python
sentence = ["the", "cat"]

grammar = {
    "S": ["the cat"]
}

text = " ".join(sentence)

if text in grammar["S"]:
    print("Accepted")
else:
    print("Rejected")
```

### 16. Implement a Python program using the SpaCy library to perform Named Entity Recognition (NER) on a given text.

```python
import spacy

nlp = spacy.load("en_core_web_sm")

doc = nlp("Virat Kohli lives in India.")

for ent in doc.ents:
    print(ent.text, ent.label_)
```

### 17. Write program demonstrates how to access WordNet, a lexical database, to retrieve synsets and explore word meanings in python.

```python
from nltk.corpus import wordnet

word = "car"

for syn in wordnet.synsets(word):
    print(syn.name(), "-", syn.definition())
```

### 19. Create a program for word sense disambiguation using the Lesk algorithm using python.

```python
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

sentence = "I went to the bank to deposit money."

sense = lesk(word_tokenize(sentence), "bank")

print(sense)
```

### 20. Implement a basic information retrieval system using TF-IDF (Term Frequency-Inverse Document Frequency) for document ranking using python.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

docs = [
    "Python is easy",
    "Python is powerful",
    "I love programming"
]

tfidf = TfidfVectorizer()

result = tfidf.fit_transform(docs)

print(result.toarray())
```
